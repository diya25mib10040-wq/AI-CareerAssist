```python
import pandas as pd
import re

# Load datasets
resume_df = pd.read_csv("../data/Resume.csv")
job_df = pd.read_csv("../data/job_dataset.csv")

# Remove extra spaces from column names
resume_df.columns = resume_df.columns.str.strip()
job_df.columns = job_df.columns.str.strip()

print("Resume dataset loaded:", resume_df.shape)
print("Job dataset loaded:", job_df.shape)

print("\nResume columns:")
print(resume_df.columns.tolist())

print("\nJob columns:")
print(job_df.columns.tolist())
```

```python
import re

def clean_text(text):
    text = str(text)
    text = text.lower()
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

print("Text cleaning function created successfully!")
```

```python
print("Resume columns:")
print(resume_df.columns.tolist())
```

```python
# Find the text column in the resume dataset
text_columns = resume_df.select_dtypes(include="object").columns

print("Text columns found:")
print(text_columns.tolist())

# Select the text column with the longest average text
resume_text_column = max(
    text_columns,
    key=lambda col: resume_df[col].astype(str).str.len().mean()
)

print("\nSelected resume text column:", resume_text_column)

# Clean the resume text
resume_df["clean_resume_text"] = resume_df[resume_text_column].apply(clean_text)

print("\nCleaned resume column created successfully!")
print(resume_df[["clean_resume_text"]].head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)

print("TF-IDF vectorizer created successfully!")
```

```python
# Convert cleaned resume text into TF-IDF vectors
resume_tfidf = tfidf_vectorizer.fit_transform(
    resume_df["clean_resume_text"]
)

print("Resume TF-IDF created successfully!")
print("Resume TF-IDF shape:", resume_tfidf.shape)
```

```python
# Clean the job responsibilities
job_df["clean_responsibilities"] = job_df["Responsibilities"].apply(clean_text)

print("Clean job responsibilities created successfully!")
print(job_df[["Responsibilities", "clean_responsibilities"]].head())
```

```python
job_tfidf = tfidf_vectorizer.transform(
    job_df["clean_responsibilities"]
)

print("Job TF-IDF created successfully!")
print("Job TF-IDF shape:", job_tfidf.shape)
```

```python
def convert_skills_to_list(skill_text):
    if pd.isna(skill_text):
        return []

    skill_text = str(skill_text).lower()

    # Replace common separators with commas
    skill_text = skill_text.replace("|", ",")
    skill_text = skill_text.replace(";", ",")
    skill_text = skill_text.replace("\n", ",")

    skills = [skill.strip() for skill in skill_text.split(",")]

    # Remove empty values
    skills = [skill for skill in skills if skill]

    return skills


print("Skill conversion function created successfully!")
```

```python
def analyze_skill_gap(resume_index, job_index):
    resume_skills = set(resume_df.iloc[resume_index]["resume_skills"])
    job_skills = set(job_df.iloc[job_index]["job_skills"])

    matched_skills = resume_skills.intersection(job_skills)
    missing_skills = job_skills.difference(resume_skills)

    return {
        "matched_skills": sorted(matched_skills),
        "missing_skills": sorted(missing_skills),
        "total_required_skills": len(job_skills),
        "total_matched_skills": len(matched_skills),
        "total_missing_skills": len(missing_skills)
    }


print("Skill gap analysis function created successfully!")
```

```python
import pandas as pd

def convert_skills_to_list(skill_text):
    if pd.isna(skill_text):
        return []

    skill_text = str(skill_text).lower()
    skill_text = skill_text.replace("|", ",")
    skill_text = skill_text.replace(";", ",")
    skill_text = skill_text.replace("\n", ",")

    skills = [skill.strip() for skill in skill_text.split(",")]
    skills = [skill for skill in skills if skill]

    return skills


# Find the resume skill column
skill_columns = [
    col for col in resume_df.columns
    if "skill" in col.lower()
]

print("Skill columns found:", skill_columns)

resume_skill_column = skill_columns[0]

resume_df["resume_skills"] = resume_df[resume_skill_column].apply(
    convert_skills_to_list
)

print("resume_skills column created successfully!")
```

```python
learning_resources = {
    "python": "Learn Python programming",
    "machine learning": "Learn Machine Learning fundamentals",
    "deep learning": "Learn Deep Learning and neural networks",
    "sql": "Learn SQL and database management",
    "data analysis": "Learn Data Analysis with Pandas and NumPy",
    "excel": "Improve Excel and data handling skills",
    "communication": "Improve communication and presentation skills",
    "project management": "Learn project management fundamentals",
    "data visualization": "Learn data visualization using Matplotlib and Seaborn",
    "statistics": "Learn statistics for data analysis",
    "html": "Learn HTML and web page structure",
    "css": "Learn CSS and web styling",
    "javascript": "Learn JavaScript programming",
    "java": "Learn Java programming",
    "c++": "Learn C++ programming",
    "git": "Learn Git and version control",
    "github": "Learn GitHub and collaborative development",
    "cloud": "Learn cloud computing fundamentals",
    "aws": "Learn AWS cloud services",
    "docker": "Learn Docker and containerization"
}


def generate_learning_recommendations(missing_skills):
    recommendations = []

    for skill in missing_skills:
        skill_lower = skill.lower().strip()

        if skill_lower in learning_resources:
            recommendations.append(
                learning_resources[skill_lower]
            )
        else:
            recommendations.append(
                f"Learn and practice {skill}"
            )

    return recommendations


print("Learning recommendation system created successfully!")
```

```python
def generate_complete_analysis(resume_index, job_index):
    # 1. Job matching
    job_result = find_best_job(resume_index)

    # 2. Skill gap analysis
    skill_result = analyze_skill_gap(resume_index, job_index)

    # 3. Career recommendation
    career_result = predict_career(resume_index)

    # 4. Learning recommendations
    learning_result = generate_learning_recommendations(
        skill_result["missing_skills"]
    )

    print("========================================")
    print("       AI CAREERASSIST ANALYSIS")
    print("========================================")

    print("\n1. BEST JOB MATCH")
    print("----------------------------------------")
    print("Job Title:", job_result["Job Title"])
    print("Experience Level:", job_result["Experience Level"])
    print("Match Score:", job_result["Match Score"], "%")

    print("\n2. SKILL GAP ANALYSIS")
    print("----------------------------------------")
    print("Matched Skills:", skill_result["matched_skills"])
    print("Missing Skills:", skill_result["missing_skills"])

    print("\n3. LEARNING RECOMMENDATIONS")
    print("----------------------------------------")

    if learning_result:
        for i, recommendation in enumerate(learning_result, start=1):
            print(f"{i}. {recommendation}")
    else:
        print("No learning recommendations required.")

    print("\n4. TOP 3 CAREER RECOMMENDATIONS")
    print("----------------------------------------")

    for rank, item in enumerate(career_result, start=1):
        print(
            f"{rank}. {item['Career']} "
            f"({item['Confidence']}%)"
        )

    print("\n========================================")

    return {
        "job_match": job_result,
        "skill_gap": skill_result,
        "learning_recommendations": learning_result,
        "career_recommendations": career_result
    }


print("Complete analysis updated successfully!")
```
